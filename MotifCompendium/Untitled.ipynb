```python
df = df[["source_models", "motifs_per_model", "contrib_per_model", "info", "chiptf_name0"]]
```

```python
df
```

```python
data = []
for i,r in df.iterrows():
    models = r["info"].split(",")
    sdict = {}
    new = []
    for j in range(len(models)):
        if models[j].split("_")[0] not in sdict:
            sdict[models[j].split("_")[0]] = float(models[j].split("_")[2])
        else:
            sdict[models[j].split("_")[0]] = max(sdict[models[j].split("_")[0]], float(models[j].split("_")[2]))

    new.append(r["chiptf_name0"] + "_" + str(i))    
    for key in ["GM12878", "HEPG2", "IMR90", "H1ESC", "K562"]:
        if key in sdict:
            new.append(sdict[key])
        else:
            new.append(0)
    data.append(new)
    
```
